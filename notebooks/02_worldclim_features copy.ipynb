## Setup

```python
# Setup
import sys
from pathlib import Path
sys.path.append(str(Path("..") / "src"))  # lets the notebook import from src/

import pandas as pd
from extract_worldclim import (
    get_raster_paths,
    get_monthly_paths,
    extract_at_points,
    process_file,
)
```

```python
# Cell 2: look at the input data
train = pd.read_csv("../data/raw/Train.csv")
print(train.shape)
train.head()
```

```python
# Cell 3: list the rasters that will be used (19 bio + elevation + 24 monthly)
paths = get_raster_paths() + get_monthly_paths()
print(len(paths), "rasters found")
[p.name for p in paths]
```

```python
# Cell 4: extract values for a small sample first
sample = train[["Latitude", "Longitude"]].head(5)
sample_feats = extract_at_points(sample, paths)
pd.concat([sample.reset_index(drop=True), sample_feats], axis=1)
```

```python
# Cell 5: check missing values across the full training file
train_feats = process_file("Train.csv", paths)
print(train_feats.shape)  # expect (number of training rows, 44)
train_feats.isna().sum().sort_values(ascending=False).head(10)
```

```python
# Cell 6: save the monthly training features
PROCESSED = Path("..") / "data" / "processed"
PROCESSED.mkdir(parents=True, exist_ok=True)
train_feats.to_csv(PROCESSED / "worldclim_monthly_train.csv", index=False)
```

```python
# Cell 7: do the same for the test set
test_feats = process_file("TestSet.csv", paths)
print(test_feats.shape)
test_feats.to_csv(PROCESSED / "worldclim_monthly_test.csv", index=False)
```
