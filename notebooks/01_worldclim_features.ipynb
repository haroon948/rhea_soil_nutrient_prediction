## Setup

```python
# setup
import sys
from pathlib import Path
sys.path.append(str(Path("..") / "src"))  # lets the notebook import from src/

import pandas as pd
from extract_worldclim import get_raster_paths, extract_at_points, process_file
```

```python
# Cell 2: look at the input data
train = pd.read_csv("../data/raw/Train.csv")
print(train.shape)
train.head()
```

```python
# Cell 3: list the WorldClim rasters that will be used
paths = get_raster_paths()
print(len(paths), "rasters found")
[p.name for p in paths]
```

```python
# Cell 4: extract values for a small sample first
sample = train[["Latitude", "Longitude"]].head(5)
sample_feats = extract_at_points(sample, paths)
pd.concat([sample.reset_index(drop=True), sample_feats], axis=1)
```

```python
# Cell 5: check missing values across the full file
train_feats = process_file("Train.csv", paths)
train_feats.isna().sum().sort_values(ascending=False).head(10)
```

```python
# Cell 6: save the results
train_feats.to_csv("../data/processed/worldclim_train.csv", index=False)
```

```python
# Cell 7: test-set features, same bio paths as training
test_feats = process_file("TestSet.csv", paths)
test_feats.to_csv("../data/processed/worldclim_test.csv", index=False)

# The two feature files must have identical columns in the same order
print(train_feats.shape, test_feats.shape)
assert list(train_feats.columns) == list(test_feats.columns), "feature columns differ"
```
